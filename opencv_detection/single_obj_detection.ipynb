```python
import cv2
from ultralytics import YOLO

#
```

```python
#  1. Initialize the YOLO v8 model
# Use 'yolov8n.pt' (Nano) for the fastest real-time performance [3, 5]
model = YOLO('yolov8n.pt')

# 2. Access the webcam feed (0 is typically the default camera) [2]
cap = cv2.VideoCapture(0)

# 3. Continuous detection loop [3]
while True:
    # Capture frame-by-frame
    ret, frame = cap.read()
    if not ret:
        break

    # 4. Run inference on the current frame [3]
    # You can specify classes=0 if you only want to detect people [6]
    results = model(frame)

    # 5. Create an annotated frame with bounding boxes and labels [3, 4]
    # results contains the detection data for the current frame
    annotated_frame = results.plot()

    # 6. Display the live feed [4]
    cv2.imshow("YOLO v8 Real-Time Detection", annotated_frame)

    # 7. Exit the loop when 'q' is pressed [4]
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 8. Release resources [4]
cap.release()
cv2.destroyAllWindows()
```
