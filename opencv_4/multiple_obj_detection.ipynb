```python
# Multiple Object Detection in Video (OpenCV + YOLO v8)
# This script processes a video file frame-by-frame, runs the YOLO model to detect all configured object classes, and displays the annotated results
# .
import cv2
from ultralytics import YOLO

```

```python
# 1. Load the pre-trained YOLO v8 Nano model
# The Nano version (yolov8n.pt) is used for efficient processing [5]
model = YOLO('yolov8n.pt')

# 2. Access the video file (provide the path to your MP4 file)
# Example: 'street.mp4' as used in the source tutorial [2]
cap = cv2.VideoCapture(0)

# 3. Process the video feed continuously
while True:
    # Capture the video frame-by-frame
    ret, frame = cap.read()
    
    # Break the loop if the video has ended or cannot be read [4]
    if not ret:
        break

    # 4. Perform detection on the current frame
    # By default, this will detect all 80 COCO classes [1, 4]
    # To detect specific classes (e.g., just people and cars), use:
    # results = model(frame, classes=[6]) 
    results = model(frame)

    # 5. Extract and plot the annotations (bounding boxes, labels, probabilities)
    # The '.plot()' method creates an image with the detection data [4, 7]
    annotated_frame = results[0].plot()

    # 6. Display the processed video frame
    cv2.imshow("Multi-Object Detection in Video", annotated_frame)

    # 7. Press 'q' to stop the video processing
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# 8. Clean up resources [3]
cap.release()
cv2.destroyAllWindows()
```
